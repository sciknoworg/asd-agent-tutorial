# 01 - Understanding the virtual lab

This notebook introduces the toy ASD simulator. It is educational and does not predict real HfO2/MoS2 chemistry.

```python
from asd_agent.config import load_scenario
from asd_agent.models import ExperimentCondition
from asd_agent.simulator import VirtualLab

config = load_scenario("inherent_selectivity").model_copy(update={"noise_sigma_nm": 0.0})
lab = VirtualLab(config, seed=1)
config.description
```

```python
condition = ExperimentCondition(
    precursor_dose_s=6.0,
    coreactant_dose_s=6.0,
    inhibitor_dose_s=0.0,
    temperature_c=180.0,
    cycles=70,
)
lab.simulate(condition, experiment_id="demo_001").model_dump()
```

```python
from asd_agent.plotting import plot_selectivity_heatmap, plot_thickness_curves

plot_thickness_curves(config, "../results/notebook_01/thickness_curves.png")
plot_selectivity_heatmap(config, "../results/notebook_01/selectivity_heatmap.png")
```
