# Bayesian Active Learning for Saturation

This notebook demonstrates the Stage 1 virtual saturation study. It compares a fixed grid, a generic GP, and a physics-informed GP with a trainable saturating mean. The model is educational and does not predict real ALD chemistry.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT / "src"))

from asd_agent.bo.study import load_stage1_study_profile, run_stage1_study, stage1_summary_rows  # noqa: E402,I001

profile = load_stage1_study_profile("smoke")
profile
```

```python
results = run_stage1_study(profile)
rows = stage1_summary_rows(profile, results)
rows[:3]
```

```python
import pandas as pd

summary = pd.DataFrame(rows)
summary[
    [
        "method",
        "scenario_id",
        "status",
        "n_experiments",
        "estimated_t95_s",
        "true_t95_s",
        "absolute_t95_error_s",
        "uncertainty_coverage",
    ]
]
```

```python
axis = summary.pivot_table(
    index="scenario_id",
    columns="method",
    values="n_experiments",
    aggfunc="mean",
).plot(kind="bar", figsize=(8, 4), title="Experiments Used in the Smoke Profile")
axis.set_ylabel("experiments")
```

The physics-informed GP uses the saturating mean as a trainable inductive bias and places a GP residual around it. It is not a fully Bayesian posterior over the physical parameters; the parameters are optimized during fitting.
